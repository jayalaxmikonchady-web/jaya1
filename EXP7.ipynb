```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
cars_df=pd.read_csv('/content/mtcars-selected-columns.csv')
__builtins__.print(cars_df)
```

```python
a=plt.scatter(cars_df['mpg'],cars_df['wt'],marker='*')
__builtins__.print()
plt.xlabel('mpg')
plt.ylabel('weight')
plt.title('Relationship between the weight of the cars and mpg')
plt.show()
```

```python
trans=cars_df['am'].value_counts()
__builtins__.print(trans)
trans.plot.bar()
plt.xlabel("tansmission")
plt.title('Bar plot to check the frequency distribution transmission type')
plt.show()
```

```python
cars_df['mpg'].plot(kind='box')
plt.xlabel('mpg')
plt.title('box and whiker plot of mpg')
plt.show()
```

```python
minvalue=np.min(cars_df['mpg'])
q1=np.percentile(cars_df['mpg'],25)
median=np.median(cars_df['mpg'])
q3=np.percentile(cars_df['mpg'],75)
maxvalue=np.max(cars_df['mpg'])
__builtins__.print('minimum value:',minvalue)
__builtins__.print('q1:',q1)
__builtins__.print('median:',median)
__builtins__.print('q3:',q3)
__builtins__.print('maxmum value:',maxvalue)
```
