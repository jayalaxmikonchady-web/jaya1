```python
import pandas as pd
import numpy as np
Series_A=pd.Series([10,20,30,40,50])
Series_B=pd.Series([40,50,60,70,80])
union=pd.Series(np.union1d(Series_A,Series_B))
intersection=pd.Series(np.intersect1d(Series_A,Series_B))
notcommonseries=union[-union.isin(intersection)]
print(notcommonseries)
```
